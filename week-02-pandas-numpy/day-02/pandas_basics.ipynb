## Pandas 

- Pandas are the Python library used to work with tabluer datasets
- Pandas are mainly work with series and dataframes datasets 
- Pandas are used data manupulation , cleaning , EDA , organise the messy data into clean format 

```python
import pandas as pd
```

```python
# Series  1-D

ages = pd.Series([25, 28, 30, 26])
ages
```

```python
# dataframes - 2D

data = {
    "name": ["Rahul", "Priya", "Amit"],
    "age": [25, 28, 30],
    "salary": [50000, 60000, 45000]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.head() # top 5 lines 
```

```python
df.tail() # bottom 5 lines 
```

```python
df.info()  #  information about the coumns like how many columns what data type of columns how many non-null vlaues 
```

```python
df.describe()  # it gives statistical summery like count ,mean , std , min , 25% , 50% , 75% , max
# it is applicable for numericals columns only
```

```python
# selecting columns 
# for one column

df['name']
```

```python
# more then 1 columns

df[['name','salary']]
```

```python
data1 = {
    'customer' : ['naresh','suresh','mahesh','tom','sam','ram','roy','rohith','virat','varun'] ,
    'age' : [23,43,54,34,33,24,56,43,45,67] ,
    'City' : ['pune','bengaluru','delhi','goa','mumbai','pune','pune','jaipure','amrithsar','bengaluru']  ,
    'salary' : [30000,25000,60000,55000,65000,54000,34000,70000,86000,100000]
}

df = pd.DataFrame(data1)
df
```

```python
# average salary
print("total salary :- ",sum(df['salary']))
print("length" , len(df['salary']))
print("avg sal :-", sum(df['salary'])/len(df['salary']))
```

```python
max(df['salary'])  # max salary 
```

```python
min(df['salary'])  # min salary 
```
