## groupby()

```python
# Total_sales by Categoeries 

df.groupby('Product_Category')['Sales_Amount'].sum()
```

```python
# Multiple statistics:

df.groupby('Product_Category')['Sales_Amount'].agg(['sum','mean','count','min','max'])
```

```python
# Multiple grouping columns:

df.groupby(['Region','Product_Category'])['Sales_Amount'].sum()
```
