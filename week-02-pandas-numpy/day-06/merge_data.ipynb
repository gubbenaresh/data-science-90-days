```python
import pandas as pd
```

```python
customer_data = {
    "customer_id": [1, 2, 3],
    "name": ["Rahul", "Priya", "Amit"],
    "city": ["Bangalore", "Pune", "Mumbai"]
}

customer = pd.DataFrame(customer_data)
customer
```

```python
order_data = {
    "order_id": [101, 102, 103],
    "customer_id": [1, 2, 1],
    "product": ["Laptop", "Mouse", "Monitor"],
    "amount": [60000, 1000, 15000]
}
order = pd.DataFrame(order_data)
order
```

```python
# using merge how we can perform inner join

merged = pd.merge(
    customer,  
    order,
    on='customer_id',
    how='inner'
)
merged
```

## Task

```python
# Total sales
# Average sales
# Sales by product
# Sales by category
# Sales by city
# Top-selling product
# Average sales per category
# Transactions per category
```

```python
df['Sales_Amount'].sum() # Total sales
```

```python
df['Sales_Amount'].mean() # Average sales
```

```python
# Sales by product

df.groupby('Product_Category')['Sales_Amount'].sum()
```

```python
# Sales by Customer_Type category

df.groupby('Customer_Type')['Sales_Amount'].sum()
```

```python
# Sales by city

df.groupby('Region')['Sales_Amount'].sum()
```

```python
# Top-selling product

df.groupby('Product_Category')['Sales_Amount'].max()
```

```python
# Average sales per category

df.groupby('Product_Category')['Sales_Amount'].mean()
```

```python
# Transactions per category

df['Product_Category'].value_counts()
```

```python
df.groupby('Product_Category').size()
```

```python
df.groupby('Product_Category')['Product_Category'].count()
```
