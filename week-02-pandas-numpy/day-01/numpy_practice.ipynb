## Creating different types of arrays

### Indexing and Slicing

```python
 # Basic Indexing

arr = np.array([10, 20, 30, 40])
print("accessing 0th index of an array:- ",arr[0])         # 10
print("accessing -1th index of an array:- ",arr[-1])        # 40
```

```python
# 2D Indexing

mat = np.array([[1, 2], [3, 4]])

print("this is 2-D array :-\n",mat)

print("="*30)

print("[0][1] 0th-> [1,2] for this 1th->[2]")
print("0 represent the 0th row for this we are accessing 1th column of it :- ",mat[0][1])     # 2
```

```python
# Slicing

print(arr)
print(arr[1:4])       # [20, 30, 40]
print(arr[:3])        # [10, 20, 30]
print(arr[::-1])      # Reverse array
```

```python
# Boolean Indexing

# we can check the values which are greater then 25 in an array 
# array[array with condition]

print(arr[arr > 25])   # Filter elements greater than 25

print(arr[arr > 55])
```

```python
# Fancy Indexing

# arr[[index,]] we can directily access the values using index 

print(arr[[0, 2]])     # Select 0th and 2nd elements
```

## Array Operations 

```python
# Arithmetic Operations

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(a + b)           # [5, 7, 9] adding two arrays
print(a * b)           # [4, 10, 18] mutipleing two array 
```

```python
# Scalar Operations
# using sinle value we can  performing operations 

# Scalar means:
# A single value.

# we can mutiple by 2 for each element in an array

print(a * 2)           # [2, 4, 6]
print(a + 5)           # here 2 and 5 are the scaler using scaler we are performing operations 


#  vectorization
# Vectorization means performing operations on the complete array without using loops. 
# vectorization is performing an operation on all elements inside an array without writing sperate code for it(explicite program)

arr = np.array([10, 20, 30])
print(arr + 5)
```

```python
# Broadcasting

# performaing an operation two different arrays with different shapes

arr = np.array([[1, 2, 3],
                [4, 5, 6]])
num = np.array([10, 20, 30])

print(arr)
print("+"*40)
print(num)
print("+"*40)
print(arr+num)
```

## Flatten & ravel

```python
# flatten and ravel 


# flatten()
# Converts multi-dimensional array into 1D array.
# Creates a copy of the data.
# If you change the original array → it does not affect the flattened array.

# ravel()
# Also converts multi-dimensional array into 1D array.
# Creates a view of the data.
# If you change the original array → it will affect the raveled array (since both share the same memory).

arr = np.array([[1, 2], [3, 4]])

flat1 = arr.flatten()  # Copy ,flatten used to convert mutiple-demenstional array to one-demenetional array
ravel1 = arr.ravel()    # View , display view of array 

arr[0][1] = 20 # changed the value

print("Original:\n", arr)
print("Flatten:\n", flat1)
print("Ravel:\n", ravel1) # ravel got changed when i changed original array it will effecting to ravel array
```

### Copy() & View()

```python
# copy()
# it will copy the content 

arr1 = np.array([[1, 2], [3, 4]])
print("original array\n",arr1)

print("+"*50)

arr2 = arr1.copy()
print("copied array arr2\n",arr2)
print("+"*50)

print("arr1 id", id(arr1))
print("arr2 id", id(arr2))
print("+"*50)

arr1[1][0] = 30 # changed arr1 
print("changes in arr1 \n", arr1)
print("not changes in arr2 \n", arr2)
```

```python
# view()
# it will view the content 

arr1 = np.array([[1, 2], [3, 4]])
print("original array\n",arr1)

print("+"*50)

arr2 = arr1.view()
print("copied array arr2\n",arr2)
print("+"*50)

print("arr1 id", id(arr1))
print("arr2 id", id(arr2))
print("+"*50)

arr1[1][0] = 300 # changed arr1 
print("changes in arr1 \n", arr1)
print("changes in arr2 \n", arr2)
```

## shallow & deep copy 

```python
import copy

# in shallow copy if inner list is also shared the value will get change 
# if outer list is only shared if we try to make changes to original array also it will not effect 

# deepcopy is when we changed the original array will not effect even it is inner or outer list is shared it will not effect for deepcopy


original = [[1, 2], [3, 4]]

normal = original
shallow = copy.copy(original)
deep = copy.deepcopy(original)

normal[0][0] = 100

print("Original :", original)  # Changed
print("Normal   :", normal)    # Changed
print("Shallow  :", shallow)   # Changed because inner list is shared
print("Deep     :", deep)      # Not changed

print("-"*50)
normal[1] = [33,44] 

print("Original :", original)  # Changed
print("Normal   :", normal)    # Changed
print("Shallow  :", shallow)   # not Changed because outer list is shared
print("Deep     :", deep)      # Not changed
```

### concatenate() and vstack() and hstack()

```python
# concaionaion used to join two variables   

a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6]])

print("original :-a\n",a)
print("-------------------------------")
print("original :-b\n",b)
# Concatenate along axis 0 (vertically)
print("-------------------------------")
print("after the concatenate axis = 0 means vetrically")
print(np.concatenate((a, b), axis=0))
print("-------------------------------")
# Concatenate along axis 1 (horizontally)
# Only works if shapes match appropriately
a = np.array([[1,1], [2,2]])
b = np.array([[3,3], [4,4]])
print("original :-a\n",a)
print("-------------------------------")
print("original :-b\n",b)
print("-------------------------------")
print("after the concatenate axis = 1 means horizontally")
print(np.concatenate((a, b), axis=1))
```

```python
# vstack is used to stttore the data in vertical
# hstack is used to stttore the data in horizontal

a = np.array([1, 2])
b = np.array([3, 4])

print("Vertical Stack:\n", np.vstack((a, b)))
print("Horizontal Stack:\n", np.hstack((a, b)))
```
