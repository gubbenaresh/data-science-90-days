### What is NumPy?

NumPy is a Python library for numerical calculations and arrays.


```python
# importing numpy 

import numpy as np
```

```python
# creating array

sales = np.array([100, 200, 300, 400, 500])
```

```python
sales
```

```python
# shape
# sales array is 1-D and it has 1 row and 5 columns
sales.shape
```

```python
# 2D array

marks = np.array([
    [80, 75, 90],
    [65, 70, 85],
    [90, 95, 88]
])

print(marks.shape) # 3rows , 3 columns 
```
