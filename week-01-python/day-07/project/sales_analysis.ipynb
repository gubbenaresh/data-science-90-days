```python
employees = [
    {"name": "Rahul", "salary": 50000, "performance": 4.5},
    {"name": "Priya", "salary": 60000, "performance": 3.8},
    {"name": "Amit", "salary": 45000, "performance": 4.8},
    {"name": "Sneha", "salary": 70000, "performance": 4.2}
]
```

```python
# program should calculate:

# Highest salary employee
# Lowest salary employee
# Average salary
# Highest performance employee
# Employees with performance >= 4
# Employees with salary > ₹50,000
# Bonus for each employee
# Final salary for each employee


```

```python
def Highest_salary_employee(employees):
    high_sal = max(employees , key=lambda x: x['salary'])
    return high_sal

Highest_salary_employee(employees)
```

```python
def Lowest_salary_employee(employees):
    high_sal = min(employees , key=lambda x: x['salary'])
    return high_sal

Lowest_salary_employee(employees)
```

```python
def Average_salary_employee(employees):
    total_sal = 0
    for i in employees:
        total_sal+=i['salary']
    print(total_sal)
    length = len(employees)
    avg_sal = total_sal/length
    print(avg_sal)

Average_salary_employee(employees)
```

```python
def Highest_performance_employee(employees):
    high_per = max(employees , key=lambda x: x['performance'])
    return high_per

Highest_performance_employee(employees)
```

```python
def performance_grater_then_4(employees):
    for i in employees:
        if i['performance']>=4:
            print(i)
performance_grater_then_4(employees)
```

```python
def salary_grater_then_50k(employees):
    for i in employees:
        if i['salary']>50000:
            print(i)
salary_grater_then_50k(employees)
```

```python
def bonus_each_employee(employees):
    for i in employees:
        i['bonus']=10000
        print(i)
bonus_each_employee(employees)
```

```python
def final_sal(employees):
    for i in employees:
        print("Final_sal",i['salary']+i['bonus'])
final_sal(employees)
```
