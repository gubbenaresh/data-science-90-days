```python
sales = [
    {"product": "Laptop", "price": 60000, "quantity": 2},
    {"product": "Mouse", "price": 1000, "quantity": 5},
    {"product": "Keyboard", "price": 2000, "quantity": 3},
    {"product": "Monitor", "price": 15000, "quantity": 2}
]
```

### Calculate Product Revenue

```python
def calculate_revenue(price, quantity):
    total_amount = price * quantity
    for i in sales:
        if total_amount == i['price']*i['quantity']:
            print(i['product'],"->",total_amount)

for i in sales:
    calculate_revenue(i['price'],i['quantity'])
```

```python
for i in sales:
    print(i['product'],"->",i['price']*i['quantity'])
```

### Total Revenue

```python
def calculate_total_revenue(sales):
    total = []
    for i in sales:
        total.append(i['price']*i['quantity'])
    print(total)
    print(sum(total))

calculate_total_revenue(sales)
```

### Highest Revenue Product

```python
def find_highest_revenue_product(sales):
    highest_revenue_product = max(sales , key=lambda x: x['price']*x['quantity'])
    print(highest_revenue_product['product'],"->",highest_revenue_product['price']*highest_revenue_product['quantity'])
find_highest_revenue_product(sales)
```

```python
total = []
for i in sales:
    total.append(i['price']*i['quantity'])
print(total)
print(max(total))
```

```python
total = []

for i in sales:
    revenue = i['price'] * i['quantity']
    total.append((i['product'], revenue))

print(total)
```

```python
highest = max(sales, key=lambda x: x['price'] * x['quantity'])

print("Highest Revenue Product:", highest['product'])
print("Revenue:", highest['price'] * highest['quantity'])
```

### average revenue

```python
def calculate_average_revenue(sales):
    total = []
    for i in sales:
        total.append(i['price']*i['quantity'])
    print(total)
    total_sales = sum(total)
    numebers_sales = len(sales)
    print(total_sales/numebers_sales)

calculate_average_revenue(sales)
```

### High-Value Products

```python
highest_value_product = max(sales , key=lambda x:x['price'])
print(highest_value_product['product'])
print(highest_value_product['price'])
```

```python
lowest_value_product = min(sales , key=lambda x:x['price'])
print(lowest_value_product['product'])
print(lowest_value_product['price'])
```
