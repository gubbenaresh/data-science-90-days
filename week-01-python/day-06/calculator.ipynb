```python
def calculate_average(numbers):
    total = sum(numbers)
    length = len(numbers)
    avg = total/length
    return avg
    
calculate_average([1,2,3,4,5])     
```

```python
def calculate_discount(price, discount):
    after_dis = price - discount
    return after_dis
calculate_discount(1000,100)
```

```python
def is_eligible(age, salary):
    if age >=18 :
        print("eligible for salary",salary)
    else:
        print("not eligible")
is_eligible(19, 25000)
```

```python
def calculate_customer_bill(price, quantity, discount):
    total_items_bill = price * quantity 
    after_dis = total_items_bill - discount 
    return after_dis
calculate_customer_bill(800,6,150)
```

```python
def analyze_numbers(numbers):
    total_numbers = len(numbers)
    avg = (sum(numbers)/total_numbers)
    min_val = min(numbers)
    max_val = max(numbers)
    return {
        "total_numbers" : total_numbers,
        "avg" : avg,
        "min" : min_val,
        "max" : max_val
    }

analyze_numbers([20,39,45,2,4,34])
```
