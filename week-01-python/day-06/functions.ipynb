What is a Function?

A function is a reusable block of code.

```python
def add(a, b):
    return a + b
```

```python
# calling function 

result = add(10, 20)

print(result)
```

```python
# parameters

def greet(name):
    print(f"Hello {name}")
```

```python
greet("naresh")
```

```python
# return is used to return the result and we can not store or we can use for future use 
add(4,5)
```
