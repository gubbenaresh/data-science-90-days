```python
sales = [100, 200, 300, 400]

print(sales[0])
```

#### List Operations 

```python
sales.append(500) # add the element in at the end of list
print(sales)
```

```python
sales.remove(400) # it used to remove specific element in the list
print(sales)
```

```python
sales.sort()
print(sales)
```

```python
len(sales)
```

```python
sum(sales)
```

### Dictionaries

```python
# store the values in keys and values format

customer = {
    "name": "Naresh",
    "age": 25,
    "city": "Bangalore",
    "salary": 60000
}
```

```python
# access

print(customer['name'])
```

```python
# add

customer['job'] = "data science"
customer
```

```python
# update 
customer['age'] = 26 
customer
```

```python
customer.keys()  # only keys 
```

```python
customer.values() # only values
```

```python
customer.items() # both
```
