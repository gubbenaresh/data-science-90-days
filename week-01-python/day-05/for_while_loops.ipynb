### for Loop 

```python
sales = [100, 200, 300, 400]

for sale in sales:
    print(sale)
```

### range()

```python
for i in range(1,5):
    print(i)
```

### Loop + Condition

```python
sales = [100, 600, 300, 800, 200]

for sale in sales:

    if sale > 500:
        print(sale)
```

### While loop

- while untile and unless the condition become false it keep itterating 

```python
# initialization 
# condition 
# increment / decrement 


count = 1

while count <= 5:
    print(count)
    count += 1
```
