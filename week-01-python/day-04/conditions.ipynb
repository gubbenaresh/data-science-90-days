```python
# if 

age = 25

if age >= 18:
    print("Adult")
```

```python
# if / else

salary = 60000

if salary >= 50000:
    print("Eligible")
else:
    print("Not Eligible")
```

```python
# elif 

marks = 75

if marks >= 90:
    print("A")
elif marks >= 75:
    print("B")
elif marks >= 60:
    print("C")
else:
    print("D")
```

```python
# nested conditions

age = 18
salary = 49000

if age >= 18:

    if salary >= 50000:
        print("Eligible")
    else:
        print("salary less then 50K")
else:
    print("age less then 18")
```
